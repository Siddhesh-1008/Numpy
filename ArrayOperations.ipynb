```python
# STACKING OF ARRAYS(HOW TO JOIN TWO ARRAYS)
# CONSISTS OF VERTICAL AND HORIZONTAL STACK
# IN STACKING US USE ONLY VECTORS AND NOT MATRIX
```

```python
# VERTICAL STACK(vertical that is columsn jodo dono array ka)
C=np.array([1,2,3,4])
D=np.array([5,6,7,8])
np.vstack((C,D))
```

```python
# HORIZONTAL STACK(horizontal that is rows jodo dono array ka)
np.hstack((C,D))
```

```python
# CoLUMN STACK INT THIS ARRANGE EACH ARRAY INTO VERTICAL MEANS VERTICAL ROTATE KARO VECTOR KO
np.column_stack((C,D))
```

```python
# SPLITTING ARRAY
# MEANS SPLITTING ARRAY INTO TWO PARTS EITHER HORIZONTAL OR VERTICAL AND SPLIT NUMBER MUST BE DIVISIBLE BY TOTAL ELEMENTS OF OROGINAL ARRAYA
```

```python
arr5=np.arange(0,16).reshape(4,4)
arr5
```

```python
np.vsplit(arr5,4)
```

```python
np.hsplit(arr5,2)
```
